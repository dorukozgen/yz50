```python
def get_stats(ids):
    counts = {}
    for pair in zip(ids, ids[1:]):
        counts[pair] = counts.get(pair, 0) + 1

    return counts

stats = get_stats(tokens)
print(sorted(((v,k) for k,v in stats.items()), reverse=True))
```

```python
top_pair = max(stats, key=stats.get)
top_pair, stats.get(top_pair)

len(tokens)
```

```python
# Merge function for Byte Pair Encoding

def merge(ids, pair, idx):
    newids = []
    i = 0
    while i < len(ids):
        if i < len(ids) - 1 and ids[i] == pair[0] and ids[i+1] == pair[1]:
            newids.append(idx)
            i += 2
        else:
            newids.append(ids[i])
            i += 1

    return newids

print(merge([1, 3, 5, 6, 6, 7, 9, 7, 6, 7, 3], (6, 7), 99))

tokens2 = merge(tokens, top_pair, 256)
print(tokens2[:100])
print('len: ', len(tokens2))
```

```python
vocab_size = 300
num_merges = vocab_size - 256
ids = list(tokens)

merges = {}
for i in range(num_merges):
    stats = get_stats(ids)
    pair = max(stats, key=stats.get)
    idx = 256 + i
    print(f"merging {pair} into a new token {idx}")
    ids = merge(ids, pair, idx)
    merges[pair] = idx
```

```python
print("tokens length:", len(tokens))
print("ids length:", len(ids))
print(f"compression ratio {len(tokens) / len(ids):.2f}X")
```

```python
vocab = {idx: bytes([idx]) for idx in range(256)}
for (p0, p1), idx in merges.items():
    vocab[idx] = vocab[p0] + vocab[p1]

def decode(ids):
    tokens = b"".join(vocab[idx] for idx in ids)
    text = tokens.decode("utf-8", errors='replace')

    return text

print(decode([196, 177]))
```

```python
def encode(text):
    tokens = list(text.encode("utf-8"))
    while True:
        stats = get_stats(tokens)
        pair = min(stats, key=lambda p: merges.get(p, float("inf")))
        if pair not in merges:
            break
        idx = merges[pair]
        tokens = merge(tokens, pair, idx)
    return tokens
        
print(encode("selam nasılsın? ıııı"))
```

```python
word_freqs = {}
for chunk in chunks:
    _tokens = tuple(chunk.encode('utf-8'))
    word_freqs[_tokens] = word_freqs.get(_tokens, 0) + 1

stats = {}
for _tokens, _freq in word_freqs.items():
    for pair in zip(_tokens, _tokens[1:]):
        stats[pair] = stats.get(pair, 0) + _freq

vocab_size = 300
num_merges = vocab_size - 256
ids = list(tokens)

merges = {}
for i in range(num_merges):
    stats = get_stats(ids)
    pair = max(stats, key=stats.get)
    idx = 256 + i
    print(f"merging {pair} into a new token {idx}")
    ids = merge(ids, pair, idx)
    merges[pair] = idx

tokens = encode(text)
print(len(tokens))
```
